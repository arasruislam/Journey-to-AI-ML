```python
from functools import reduce

with open('../assets/files/sample.txt', 'r') as file:
  string_list = file.readlines()

  # total number count
  total_lines = len(string_list)
  print("total Line: ", total_lines)

  # total word count
  # file.seek(0)
  # total_word = 0;
  # for line in file:
  #   count = line.split()
  #   total_word += len(count)
  # print(total_word)

  number_of_words = list(map(lambda x: len(x.split()), string_list))
  total_words = reduce(lambda x, y: x+y, number_of_words)
  print("total word: ",total_words)

  # Numbers of Characters
  # remove new lines
  string_list = list(map(str.strip, string_list))
  # print(string_list)

  # remove space
  string_list = list(map(lambda x: x.replace(" ", ""), string_list))
  # print(string_list)

  # character
  number_of_character = list(map(lambda x: len(x), string_list))
  # print(number_of_character)
  total_number_of_character = reduce(lambda x,y: x+y, number_of_character)
  print("total character: ",total_number_of_character)

with open('../assets/files/counter_of_string.txt', 'w') as file:
  file.write(f"Total Line: {total_lines}\nTotal number of word: {total_words}\nTotal number or character: {total_number_of_character}")
```
