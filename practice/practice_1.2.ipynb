```python
try:
  def create_file(filename, content):
    with open(f"../assets/{filename}", "w+") as file:
      file.write(content)
      file.seek(0)
      print(file.readline())

  create_file("notes.txt", "Hello, this is my first file")
except Exception as error:
  print(error)
finally:
  print("Execution Done.")

```

```python
def read_line(filename):
  with open(f"../assets/files/{filename}", "r") as file:
    lines = file.readlines()

    for i, v in enumerate(lines, 1):
      print(f"{i}: {v.strip()}")

try:
  values = read_line("fruits.txt")
except Exception as err:
  print(err)
finally:
  print("Execution Done.")
```

```python
def add_log(message):
  with open("../assets/files/log.txt", "a+") as file:
    file.write(message + "/n")

add_log("User logged in")
add_log("File uploaded")
```

```python
def read_from_position(filename, position):
  with open(f"../assets/files/{filename}", "w+") as file:
    file.write("Hello, World!")
    file.seek(position)
    return file.readline()

try:
  result = read_from_position("hello.txt", 7)
  print(result)
except Exception as err:
  print(err)
```

```python
def safe_read(filename):
    try:
      with open(f"../assets/files/{filename}", "r") as file:
        content = file.read()

    except FileNotFoundError:
      print("Error: File not found!")

    else:
      print(content)

    finally:
      print("Operation complete.")

safe_read("missing.txt")
safe_read("hello.txt")
```
