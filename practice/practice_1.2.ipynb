```python
try:
  def create_file(filename, content):
    with open(f"../assets/{filename}", "w+") as file:
      file.write(content)
      file.seek(0)
      print(file.readline())

  create_file("notes.txt", "Hello, this is my first file")
except Exception as error:
  print(error)
finally:
  print("Execution Done.")

```

```python
def read_line(filename):
  with open(f"../assets/files/{filename}", "r") as file:
    return file.readlines()

try:
  values = read_line("fruits.txt")
  for i in range(0, len(values)):
    print(f"{i}: {values[i]}", end='')
except Exception as err:
  print(err)
finally:
  print("Execution Done.")
```
